```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Array:", a)
print("First element:", a[0])
print("Last element:", a[-1])
```

```python
import numpy as np

a = np.array([10, 20, 30])
b = np.array([1, 2, 3])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(a))
print("Mean:", np.mean(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))
```

```python
import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6]])

print("Matrix:")
print(a)

print("Shape:", a.shape)
print("Element at row 1, column 2:", a[0, 1])
```

```python
import pandas as pd

data = {
    'Name': ['Ravi', 'Priya', 'Arun'],
    'Age': [20, 21, 19],
    'Marks': [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

```python
import pandas as pd

data = {
    'Marks': [80, 90, 70, 85, 95]
}

df = pd.DataFrame(data)

print("Sum:", df['Marks'].sum())
print("Average:", df['Marks'].mean())
print("Maximum:", df['Marks'].max())
print("Minimum:", df['Marks'].min())
```
