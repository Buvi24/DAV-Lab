data preprocessing- handling missing data using imputation (mean, median and forward fill) and encoding categorical variables.

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.DataFrame({
    "Marks":[80,None,90,70,],
    "Attendance":[90,None,85,62],
    'Department':['CSE','ISE','ECE',None]
})
#mean imputation
df["Marks"]=df["Marks"].fillna(df["Marks"].mean())
#median imputation
df["Attendance"]=df["Attendance"].fillna(df["Attendance"].median())
#Forward fill
df["Attendance"]=df["Attendance"].ffill()
#fill missing categorical value using mode
df["Department"]=df["Department"].fillna(df["Department"].mode()[0])
#label encoding
encoder=LabelEncoder()
df['Department']=encoder.fit_transform(df['Department'])
print(df)
```
