```python
df=pd.read_csv(r"diabetes_dataset.csv")
df.head()
```

```python
df.columns
```

```python
pit.hist(df['Glucose'])
pit.xlabel('Glucose')
pit.ylabel('Frequency')
pit.title("Distribution of Glucose")
pit.show()
```

```python
pit.hist(df['BMI'],bins=10)
pit.xlabel('BMI')
pit.ylabel('Frequency')
pit.title('Distribution of BMI')
```

```python
pit.boxplot(df['Glucose'])
pit.ylabel('Glucose')
pit.title('Box Plot of Glucose')
pit.show()
```

```python
pit.scatter(df['Age'],df['Glucose'])
pit.xlabel('Age')
pit.ylabel('Glucose')
pit.title('Age vs Glucose Levels')
pit.show()
```
