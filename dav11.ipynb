```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import make_blobs
from sklearn.cluster import AgglomerativeClustering

# 1. Generate synthetic 2D data
X, y = make_blobs(n_samples=50, centers=3, random_state=42, cluster_std=1.0)

# Visualize the initial data points
plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1], s=50, cmap='viridis')
plt.title("Generated Data")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```

```python
# 2. Perform hierarchical linkage
linked = linkage(X, method='ward')

# 3. Plot Dendrogram
plt.figure(figsize=(10, 6))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=True)
plt.title("Hierarchical Clustering Dendrogram (Ward Linkage)")
plt.xlabel("Sample Index")
plt.ylabel("Distance")
plt.show()
```

```python
# 4. Apply Agglomerative Clustering
n_clusters = 3
hierarchical_model = AgglomerativeClustering(n_clusters=n_clusters, metric='euclidean', linkage='ward')
labels = hierarchical_model.fit_predict(X)

# 5. Plot the clustered results
plt.figure(figsize=(6, 4))
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='rainbow', s=50)
plt.title(f"Agglomerative Clustering Results (k={n_clusters})")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.show()
```
